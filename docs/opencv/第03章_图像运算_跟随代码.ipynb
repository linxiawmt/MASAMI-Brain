## 图像的加权求和
```python
dst=cv2.addWeighted(src1,alpha,src2,beta,gamma)
```
$dst=saturate(src1*\alpha+src2*\beta+\gamma)$

````python
import cv2
import numpy as np

img1=np.ones((512,512),dtype=np.uint8)*100
img2=np.ones((512,512),dtype=np.uint8)*10

gamma=3
img3=cv2.addWeighted(img1,0.7,img2,2,gamma)
cv_show('img3',img3)
print(img3)
````

## 按位逻辑运算

### 常见的位运算函数
- cv2.bitwise_and()：与运算
- cv2.bitwise_or()：或运算
- cv2.bitwise_xor()：异或运算
- cv2.bitwise_not()：取反运算
- cv2.bitwise_not()：位与运算

````python
import cv2
import numpy as np

a=np.random.randint(0,255,(10,10),dtype=np.uint8)
b=np.zeros((10,10),dtype=np.uint8)
b[0:3,0:3]=255
b[4,4]=255

c=cv2.bitwise_and(a,b)

print("a=\n",a)
print("b=\n",b)
print("c=\n",c)

````

````python
#构造掩模图像
import cv2
import numpy as np

img=cv2.imread("picture2.png")
shape=img.shape
mask=np.zeros(shape,dtype=np.uint8)

mask[100:300,100:300]=255
mask[200:400,200:400]=255

img_and=cv2.bitwise_and(img,mask)
img_or=cv2.bitwise_or(img,mask)
img_not=cv2.bitwise_not(img)
img_xor=cv2.bitwise_xor(img,mask)

cv_show("img",img)
cv_show("mask",mask)
cv_show("img_and",img_and)
cv_show("img_or",img_or)
cv_show("img_not",img_not)
cv_show("img_xor",img_xor)
````

## 掩模
掩模是一种特殊的图像，它是由0和1组成的二值图像，用来对原始图像进行选择性的操作

当使用掩模对图像进行操作时，只会对掩模中的1所对应的区域进行操作，其他区域不进行操作。

如cv2.add(src1, src2, mask)

````python
img1=np.ones((4,4),dtype=np.uint8)*5
img2=np.ones((4,4),dtype=np.uint8)*2
mask=np.zeros((4,4),dtype=np.uint8)
mask[1:3,1:3]=1

img3=cv2.add(img1,img2,mask=mask)
print("img3=\n",img3)
````

````python
img=cv2.imread("picture2.png")

mask=np.zeros(img.shape[:2],np.uint8)

mask[100:300,100:300]=255
mask[500:700,500:700]=255

img=cv2.bitwise_and(img,img,mask=mask)

cv_show("img",img)
````

## 图像与数值的运算
## 位平面分解
将灰度图像处于同一个比特位上的二进制像素点组分别合成一副二进制值的图像，称为位平面分解。
- 8位灰度图中，有$values=a_7\times 2^7+a_6\times 2^6+a_5\times 2^5+a_4\times 2^4+a_3\times 2^3+a_2\times 2^2+a_1\times 2^1+a_0\times 2^0$

````python
import cv2
import numpy as np
#图像预处理
img=cv2.imread("picture1.jpg",cv2.IMREAD_GRAYSCALE)
m,n=img.shape
#构造提取矩阵
kernel=np.zeros((m,n,8),dtype=np.uint8)
for i in range(8):
    kernel[:,:,i]=2**(i)
#提取位平面
result=np.zeros((m,n,8),dtype=np.uint8)
for i in range (8):
    result[:,:,i]=cv2.bitwise_and(img,kernel[:,:,i])
    cv_show("result"+str(i),result[:,:,i])
#阈值处理:二值化
for i in range(m):
    for j in range(n):
        for k in range(8):
            if result[i,j,k]>0:
                result[i,j,k]=255
            else:
                result[i,j,k]=0
for i in range(8):
    cv_show("result"+str(i),result[:,:,i])
#显示图像
````

## 数字水印
- 嵌入过程
- 提取过程

````python
#最简单的图像嵌入和提取
watermark=cv2.imread('watermark.jpg',0)
cv_show('watermark',watermark)
w=watermark[:,:]>0
watermark[w]=1
cv_show('watermark',watermark)

img=cv2.imread("picture1.jpg",cv2.IMREAD_GRAYSCALE)

kernel=np.ones(img.shape,np.uint8)*254
img=cv2.bitwise_and(img,kernel)

cv_show('img',img)
#嵌入过程
img=cv2.bitwise_or(img,watermark)
cv_show('img',img)
#提取过程
kernel=np.ones(watermark.shape,np.uint8)
extract=cv2.bitwise_and(img,kernel)
cv_show('extract',extract)
h=extract[:,:]>0
extract[h]=255
cv_show('extract',extract)
````
