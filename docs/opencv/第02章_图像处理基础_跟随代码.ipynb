
## 使用numpy.array访问像素
两个函数通过优化处理比索引访问更快：
- item()方法可以访问像素值
- itemset()方法可以修改像素值   但是已经被弃用，不推荐使用

````python
import cv2
import numpy as np

def cv_show(name,img):
    cv2.imshow(name,img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

img=np.random.randint(0,255,size=(100,100),dtype=np.uint8)
cv_show("img",img)
print("读取像素点:img.item(50,50)=",img.item(50,50))
img[50,50]=255
print("修改像素点:img.item(50,50)=",img.item(50,50))
````

## 通道处理
- 通道拆分
- 通道合并

````python
#通道拆分
img=cv2.imread("picture1.jpg")
b,g,r=cv2.split(img)
cv_show("Blue",b)
cv_show("Green",g)
cv_show("Red",r)

img1=img.copy()
img1[:,:,0]=0
cv_show("Blue",img1)
img2=img.copy()
img2[:,:,1]=0
cv_show("Green",img2)
img3=img.copy()
img3[:,:,2]=0
cv_show("Red",img3)
````

````python
#通道合并
img=cv2.merge((b,g,r))
cv_show('Merged1',img)
img=cv2.merge([r,g,b])
cv_show('Merged2',img)
img=cv2.merge((g,b,r))
cv_show('Merged3',img)
````
