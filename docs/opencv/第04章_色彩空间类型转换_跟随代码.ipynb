# 色彩空间类型转换
## 色彩空间基础
- GRAY
- RGB
- XYZ:便于计算机处理的颜色空间
- YCrCb：强调了亮度信息
- HSV：色调、饱和度、亮度信息
- HLS：色调、光亮度、饱和度信息
- CIEL\*a\*b\*
- CIEL\*u\*v\*
- Bayer

# 类型转换函数
```python
dst=cv2.cvtColor(src,code)
```
- src: 输入图像
- code: 目标图像的色彩空间和深度，可以是以下值：
    - cv2.COLOR_BGR2GRAY: 转换为灰度图像
    - cv2.COLOR_BGR2HSV: 转换为HSV色彩空间
    - cv2.COLOR_BGR2RGB: 转换为RGB色彩空间
    - cv2.COLOR_HSV2BGR: 转换为BGR色彩空间
    - cv2.COLOR_HSV2RGB: 转换为RGB色彩空间
    - cv2.COLOR_RGB2BGR: 转换为BGR色彩空间
    - cv2.COLOR_RGB2HSV: 转换为HSV色彩空间

````python
import cv2
import numpy as np
#BGR转灰度
img=np.random.randint(0,256,size=(2,4,3),dtype=np.uint8)
print("img=\n",img)

rst=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
print("rst=\n",rst)
````

## HSV色彩空间

H:色调，取值范围0-360，0表示红色，120表示绿色，180表示青色，240表示蓝色,300表示品红色。

S:饱和度，取值范围0-255，0表示灰色，255表示饱和度最高。

V:明度，取值范围0-255，0表示黑色，255表示白色。

### 获取指定颜色


````python
import cv2
import numpy as np
#===================蓝色图像======
imgBlue=np.zeros([1,1,3],np.uint8)
imgBlue[0,0,0]=255
Blue=imgBlue.copy()

BlueHSV=cv2.cvtColor(Blue,cv2.COLOR_BGR2HSV)
print(Blue)
print(BlueHSV)
#===================绿色图像======
imgGreen=np.zeros([1,1,3],np.uint8)
imgGreen[0,0,1]=255
Green=imgGreen.copy()

GreenHSV=cv2.cvtColor(Green,cv2.COLOR_BGR2HSV)
print(Green)
print(GreenHSV)
#===================红色图像======
imgRed=np.zeros([1,1,3],np.uint8)
imgRed[0,0,2]=255
Red=imgRed.copy()

RedHSV=cv2.cvtColor(Red,cv2.COLOR_BGR2HSV)
print(Red)
print(RedHSV)
````

````python
#显示特定颜色
import cv2
import numpy as np

#读取照片
img=cv2.imread("picture4.jpg")
cv_show("original",img)
#转化
hsv=cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
#====确定蓝色范围====
mask=cv2.inRange(hsv,(100,100,100),(130,255,255))

blue=cv2.bitwise_and(img,img,mask=mask)
cv_show("blue",blue)
#====确定绿色范围====
mask=cv2.inRange(hsv,(50,100,100),(70,255,255))
green=cv2.bitwise_and(img,img,mask=mask)
cv_show("green",green)
#====确定红色范围====
mask=cv2.inRange(hsv,(0,100,100),(10,255,255))
red=cv2.bitwise_and(img,img,mask=mask)
cv_show("red",red)
````

### 标记肤色

````python
img=cv2.imread("picture3.jpg")
hsv=cv2.cvtColor(img,cv2.COLOR_BGR2HSV)

h,s,v=cv2.split(hsv)

minh=5;maxh=170
mins=25;maxs=166
hueMask=cv2.inRange(h,minh,maxh)
satMask=cv2.inRange(s,mins,maxs)
mask=hueMask&satMask

roi=cv2.bitwise_and(img,img,mask=mask)
cv_show("roi",roi)
````

### 实现艺术效果

````python
img = cv2.imread('picture1.jpg')
hsv=cv2.cvtColor(img,cv2.COLOR_BGR2HSV)

h,s,v=cv2.split(hsv)
v[:,:]=120

result=cv2.merge([h,s,v])

result=cv2.cvtColor(result,cv2.COLOR_HSV2BGR)

cv_show('result',result)
````

## alpha通道
- 透明度
- 在显示中不同alpha值的图像没有差别
- 用imwrite函数保存图像可以得到不同透明度的图像，这时有区别

````python
img=cv2.imread("picture3.jpg")
img_bgra=cv2.cvtColor(img,cv2.COLOR_BGR2BGRA)
print(img_bgra)
b,g,r,a=cv2.split(img_bgra)
a[:,:]=125
img_bgra=cv2.merge((b,g,r,a))
print(img_bgra)
cv_show("BGRA",img_bgra)
cv2.imwrite("picture3_bgra.png",img_bgra)
````
