## VideoCapture类

### 摄像头

```python
cap=cv2.VideoCapture(0) # 打开摄像头
if cap.isOpened():
    open,frame=cap.read()
else:
    open=False
while open:
    open,frame=cap.read()
    if frame is None:
        break
    if open==True:
        cv2.imshow("camera",frame)
        c=cv2.waitKey(1)
        if c==27:
            break
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

# 打开摄像头（0 表示默认摄像头）
cap = cv2.VideoCapture(0)

# 检查是否成功打开
if not cap.isOpened():
    print("无法打开摄像头")
    exit()

while True:
    # 逐帧捕获
    ret, frame = cap.read()
    
    # 如果读取失败，则退出
    if not ret:
        print("无法接收帧（可能摄像头断开）")
        break
    
    # 显示当前帧
    cv2.imshow('Camera', frame)
    
    # 按下 'q' 键退出
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# 释放摄像头并关闭窗口
cap.release()
cv2.destroyAllWindows()
```

### 播放视频文件


```python
cap=cv2.VideoCapture('video1.mp4')

while(cap.isOpened()):
    ret,frame=cap.read()
    if frame is None:
        break
    if ret==True:
        cv2.imshow("frame",frame)
        c=cv2.waitKey(25)
        if c==27:
            break

cap.release()
cv2.destroyAllWindows()
```

## VideoWriter类

```python
import cv2
import numpy as np

cap=cv2.VideoCapture(0)

fourcc=cv2.VideoWriter_fourcc('I','4','2','0')
out=cv2.VideoWriter("output.avi",fourcc,30,(640,480),True)
while(cap.isOpened()):
    ret,frame=cap.read()
    if frame is None:
        break
    if ret==True:
        out.write(frame)
        cv2.imshow("frame",frame)
        if cv2.waitKey(1)==27:
            break
    else:
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```

## 视频操作基础


```python
cap=cv2.VideoCapture(0) #打开摄像头

while(cap.isOpened()):
    ret,frame=cap.read() #读取一帧图像
    if frame is None:
        break
    if ret is True:
        edges=cv2.Canny(frame,50,150)
        cv2.imshow('frame',edges) #显示图像
        if cv2.waitKey(1)==27:
            break

cap.release()
cv2.destroyAllWindows()
```

```python
cap=cv2.VideoCapture(0) #打开摄像头

while(cap.isOpened()):
    ret,frame=cap.read() #读取一帧图像
    if frame is None:
        break
    if ret is True:
        gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY) #转换为灰度图
        kernel=np.ones((5,5),np.uint8) #定义卷积核
        erosion=cv2.erode(gray,kernel,iterations=1) #腐蚀
        b=cv2.subtract(gray,erosion) #差分
        cv2.imshow('frame',b) #显示图像
        if cv2.waitKey(1)==27:
            break

cap.release()
cv2.destroyAllWindows()
```
