## 腐蚀

```python
def cv_show(name,img):
    cv2.imshow(name,img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
import cv2
import numpy as np

import cv2
import numpy as np

img=np.zeros((5,5),np.uint8)

img[1:4,1:4]=1
kernel=np.ones((2,2),np.uint8)
erosion=cv2.erode(img,kernel,iterations=1)
print("img=\n", img)
print("kernel=\n", kernel)
print("erosion=\n", erosion)

img=cv2.imread("picture2.png",cv2.IMREAD_GRAYSCALE)

#自适应阈值处理-高斯
img=cv2.adaptiveThreshold(img,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,5,3)

kernel=np.ones((2,1),np.uint8)
erosion=cv2.erode(img,kernel,iterations=1)
cv_show("img",img)
cv_show("erosion",erosion)
```

## 膨胀


```python
kernel=np.ones((2,1),np.uint8)
dilation=cv2.dilate(img,kernel,iterations=1)
cv_show('Original',img)
cv_show('Dilation',dilation)
```

## 通用形态学函数
`dst=cv2.morphologyEx(src,op,kernel,anchor,iterations,borderType,borderValue)`
- `src`：输入图像，必须为二值图像。
- `op`：形态学操作类型，可以是以下值：
  - `MORPH_ERODE`：腐蚀操作。
  - `MORPH_DILATE`：膨胀操作。
  - `MORPH_OPEN`：开运算，先腐蚀再膨胀。
  - `MORPH_CLOSE`：闭运算，先膨胀再腐蚀。
  - `MORPH_GRADIENT`：梯度运算，输出图像中每个像素的梯度值。
  - `MORPH_TOPHAT`：顶帽运算，输出图像中每个像素的顶帽值。
  - `MORPH_BLACKHAT`：黑帽运算，输出图像中每个像素的黑帽值。
- `kernel`：形态学核，必须是奇数尺寸的矩形结构元素。
- `anchor`：锚点，指定锚点位置，默认值为(-1,-1)，表示锚点位于核的中心。
- `iterations`：迭代次数，默认值为1。
- `borderType`：边界类型，可以是以下值：
  - `BORDER_CONSTANT`：常值边界，常用值是0。
  - `BORDER_REPLICATE`：复制边界，将边界像素值复制到相邻的边界上。
  - `BORDER_REFLECT`：反射边界，将边界像素值反射到相邻的边界上。
  - `BORDER_WRAP`：环绕边界，将边界像素值绕到另一边。
  - `BORDER_REFLECT_101`：101反射边界，将边界像素值反射到相邻的边界上，但不包括中心像素。
- `borderValue`：边界值，当边界类型为`BORDER_CONSTANT`时，指定常值边界值。
- `dst`：输出图像。

### 开运算：先腐蚀再膨胀


```python
kernel=np.ones((2,1),np.uint8)
opening=cv2.morphologyEx(img,cv2.MORPH_OPEN,kernel,iterations=1)
cv_show('img',img)
cv_show('opening',opening)


erosion=cv2.erode(img,kernel,iterations=1)
dilation=cv2.dilate(erosion,kernel,iterations=1)
cv_show("dilation",dilation)
```

### 闭运算：先膨胀后腐蚀

```python
closing=cv2.morphologyEx(img,cv2.MORPH_CLOSE,kernel)
cv_show('img',img)
cv_show('closing',closing)
```

### 形态学梯度运算：图像的膨胀图像减腐蚀图像


```python
result=cv2.morphologyEx(img,cv2.MORPH_GRADIENT,kernel)
cv_show('img',img)
cv_show('result',result)

result1=dilation-erosion
cv_show('result',result1)
```

### 礼帽运算：用原始图像减去开运算图像的操作，可以获得噪声和边缘信息

```python
#噪声
result=cv2.morphologyEx(img,cv2.MORPH_TOPHAT,kernel,iterations=1)
cv_show('result',result)

result=img-opening
cv_show('result',result)

#边缘检测
img_1=cv2.imread("picture1.jpg",cv2.IMREAD_GRAYSCALE)
cv_show('img_1',img_1)

result=cv2.morphologyEx(img_1,cv2.MORPH_TOPHAT,kernel,iterations=1)
cv_show('result',result)
```

### 黑帽运算：闭运算图像减去原始图像

作用：消除噪声、去除小物体、提取边缘、提取轮廓等。

```python
result=cv2.morphologyEx(img, cv2.MORPH_BLACKHAT, kernel)
cv_show('result', result)
```

### 核函数

作用：构造形态学操作需要的kernel，用于处理图像中的像素。
`retval=cv2.getStructuringElement(shape, ksize, anchor=(-1,-1))`
- shape：结构元素的形状，可以是矩形、椭圆、十字形等。
  - cv2.MORPH_RECT：矩形结构元素。
  - cv2.MORPH_ELLIPSE：椭圆结构元素。
  - cv2.MORPH_CROSS：十字形结构元素。
- ksize：结构元素的大小，是一个奇数。
- anchor：锚点，默认值为(-1,-1)，表示锚点位于中心。

```python
kernel_1=cv2.getStructuringElement(cv2.MORPH_RECT,(59,59))
kernel_2=cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(59,59))
kernel_3=cv2.getStructuringElement(cv2.MORPH_CROSS,(59,59))
print("kernel_1:\n",kernel_1)
print("kernel_2:\n",kernel_2)
print("kernel_3:\n",kernel_3)
```

```python
import cv2

img=cv2.imread("picture7.png",cv2.IMREAD_GRAYSCALE)

dst1=cv2.dilate(img,kernel_1)
dst2=cv2.dilate(img,kernel_2)
dst3=cv2.dilate(img,kernel_3)

cv_show("img",img)
cv_show("dst1",dst1)
cv_show("dst2",dst2)
cv_show("dst3",dst3)
```
