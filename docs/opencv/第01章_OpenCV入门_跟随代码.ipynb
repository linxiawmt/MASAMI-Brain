## 图像处理基本操作
- 读取图像 cv2.imread(filenames[,flags])
- 显示图像 cv2.imshow()
- 保存图像 cv2.imwrite()

```python
import cv2
img=cv2.imread("picture1.jpg")
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
def cv_show(name,img):
    cv2.imshow(name,img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

```

```python
#waitKey(0)  # 0表示无限等待，单位是毫秒cv2.imshow("Image",img)
img=cv2.imread("picture1.jpg")
cv2.imshow("Image",img)
key=cv2.waitKey()
if key==ord('A'):
    cv2.destroyAllWindows()
    cv_show("Image",img)
elif key==ord('B'):
    cv2.destroyAllWindows()
    img2=cv2.imread("picture1_gray.jpg")
    cv_show("Image2",img2)
```

```python
img=cv2.imread("picture1.jpg",cv2.IMREAD_GRAYSCALE)
cv_show("Gray Image",img)
cv2.waitKey()                        
cv2.destroyAllWindows()
cv2.imwrite("picture1_gray.jpg",img)
```
