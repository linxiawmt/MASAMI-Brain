### threshold函数

```python
retal,dst=cv2.threshold(src,thresh,maxval,type)
```

- `src`：输入图像，必须是单通道的灰度图像。
- `thresh`：阈值，图像的像素值大于该阈值时，被认为是白色，小于该阈值时，被认为是黑色。
- `maxval`：最大值，图像的像素值大于等于该值时，被认为是白色，小于该值时，被认为是黑色。
- `type`：阈值类型，有三种：
  - `cv2.THRESH_BINARY`：二值化阈值，图像的像素值大于阈值时，被认为是白色，小于阈值时，被认为是黑色。
  - `cv2.THRESH_BINARY_INV`：反二值化阈值，图像的像素值大于阈值时，被认为是黑色，小于阈值时，被认为是白色。
  - `cv2.THRESH_TRUNC`：截断阈值，图像的像素值大于阈值时，被截断为阈值，小于阈值时，保持不变。

````python
#===二值化阈值处理  cv2.THRESH_BINARY=====

import cv2
import numpy as np

img=np.random.randint(0,256,size=(5,5),dtype=np.uint8)

thresh=128
maxval=255

retval,dst=cv2.threshold(img,thresh,maxval,cv2.THRESH_BINARY)
print(img)
print(dst)
print(retval)

img=cv2.imread("picture3.jpg")
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

thresh=127
maxval=255

retval,dst=cv2.threshold(gray,thresh,maxval,cv2.THRESH_BINARY)
cv_show("binary",dst)
````

````python
#===反二值化阈值处理    cv2.THRESH_BINARY_INV===#

retval,dst=cv2.threshold(img,thresh,maxval,cv2.THRESH_BINARY_INV)

cv_show("binary_inv",dst)
print(dst)

thresh=180
maxval=255
img=cv2.imread("picture2.png")
retval,dst=cv2.threshold(img,thresh,maxval,cv2.THRESH_BINARY_INV)

cv_show("binary_inv",dst)
````

````python
#===截断阈值化处理   cv2.THRESH_TRUNC===#
img=cv2.imread("picture3.jpg")
thresh=155
retval,dst=cv2.threshold(img,thresh,255,cv2.THRESH_TRUNC)
cv_show('dst',dst)
````

### 自适应阈值处理
作用：通过对图像的局部进行统计分析，确定阈值，从而对图像进行二值化处理。

`cv2.adaptiveThreshold(src, maxValue, adaptiveMethod, thresholdType, blockSize, C)`
- src：输入图像，单通道或三通道。
- maxValue：最大阈值。
- adaptiveMethod：自适应阈值处理方法，有三种：
  - cv2.ADAPTIVE_THRESH_MEAN_C：计算邻域的平均值，作为阈值。
  - cv2.ADAPTIVE_THRESH_GAUSSIAN_C：计算邻域的高斯加权平均值，作为阈值。
  - cv2.ADAPTIVE_THRESH_MEAN_C + cv2.THRESH_BINARY：先进行均值处理，再进行二值化处理。
- thresholdType：阈值类型，有两种：
  - cv2.THRESH_BINARY：二值化。
  - cv2.THRESH_BINARY_INV：反二值化。
- blockSize：邻域大小。
- C：常数值，用于计算阈值。

````python
import cv2

img=cv2.imread("picture1.jpg")
cv_show("Original Image",img)

img=cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)

t1,thd=cv2.threshold(img,127,255,cv2.THRESH_BINARY)

ahtdmean=cv2.adaptiveThreshold(img,255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,5,3)

athdgaus=cv2.adaptiveThreshold(img,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,5,3)

cv_show("Original Image",img)
cv_show("Binary Threshold",thd)
cv_show("Adaptive Mean Threshold",ahtdmean)
cv_show("Adaptive Gaussian Threshold",athdgaus)
````

### Ostu处理
作用：对图像进行局部自适应阈值分割，提取图像中的明显的边缘和噪声。

````python
img=cv2.imread("picture2.png")
img=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
cv_show("Original",img)

t,ostu=cv2.threshold(img,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)
print(t)
cv_show("OSTU",ostu)
````
