# 第五章 几何变换
- 缩放：cv2.resize(src,dsize,fx,fy,interpolation)


当缩小图像时，使用区域插值方式（INTER_AREA）能够得到最好的效果；当放大图像时，
使用三次样条插值（INTER_CUBIC）方式和双线性插值（INTER_LINEAR）方式都能够取得
较好的效果。三次样条插值方式速度较慢，双线性插值方式速度相对较快且效果并不逊色。
- 翻转:cv2.flip(src,flipCode)
- 仿射:cv2.warpAffine(src,M,dsize,flags,borderMode,borderValue)
$$仿射图像R=变换矩阵\times 原始图像O
$$
        - M: 2x3的变换矩阵，包括旋转、缩放、平移等变换
        - dsize: 输出图像的大小
        - flags: 插值方法
        - borderMode: 边界填充方式
        - borderValue: 边界填充值
        - 旋转仿射获取矩阵:cv2.getRotationMatrix2D(center,angle,scale)
        - 复杂仿射获取矩阵：cv2.getAffineTransform(src,dst)
- 透视
- 重映射

```python
#缩放
import cv2
import numpy as np

img=cv2.imread("picture3.jpg")
#====通过参数dsize来设置缩放大小====
dst=cv2.resize(img,(640,480))
cv_show("original",img)
cv_show("dst",dst)
##====通过参数fx和fy来设置缩放比例====
dst=cv2.resize(img,None,fx=0.5,fy=0.5)
cv_show("original",img)
cv_show("dst",dst)
```

```python
#翻转
import cv2  
import numpy as np  

img=cv2.imread("picture1.jpg")
x=cv2.flip(img,0) #flipcode=0,绕x轴翻转
y=cv2.flip(img,1) #flipcode>0,绕y轴翻转
xy=cv2.flip(img,-1) #flipcode<0,绕x轴和y轴同时翻转

cv_show("Original",img)
cv_show("Flip X",x)
cv_show("Flip Y",y)
cv_show("Flip XY",xy)
```

```python
#仿射


#===平移===#
img = cv2.imread("picture3.jpg")
h,w=img.shape[:2]
x=100
y=50
M=np.float32([[1,0,x],[0,1,y]])

move=cv2.warpAffine(img,M,(w,h),borderValue=(255,255,255))
cv_show("move",move)
move_reflect=cv2.warpAffine(img,M,(w,h),flags=cv2.WARP_INVERSE_MAP,borderValue=(255,255,255))
cv_show("move_reflect",move_reflect)

#===旋转===#

#获取转换矩阵
retal=cv2.getRotationMatrix2D((w/2,h/2),45,0.7)#center:旋转中心,angle:旋转角度,scale:缩放因子
#进行旋转
rotate=cv2.warpAffine(img,retal,(w,h))
cv_show("rotate",rotate)


#===更复杂的仿射变换===#

rows,cols,ch=img.shape

p1=np.float32([[0,0],[cols-1,0],[0,rows-1]])
p2=np.float32([[0,rows*0.3],[cols*0.8,rows*0.2],[cols*0.1,rows*0.7]])

M=cv2.getAffineTransform(p1,p2)#输入前后两组三个点的坐标，输出仿射变换矩阵
transform=cv2.warpAffine(img,M,(cols,rows))
cv_show("transform",transform)
```

```python
#透视
img=cv2.imread('picture3.jpg')
rows,cols=img.shape[:2]
pts1=np.float32([[150,50],[400,50],[60,450],[310,450]])
pts2=np.float32([[150,60],[rows-50,50],[50,cols-50],[rows-50,cols-50]])

M=cv2.getPerspectiveTransform(pts1,pts2)

dst=cv2.warpPerspective(img,M,(rows,cols))

cv_show('original',img)
cv_show('perspective',dst)
```
