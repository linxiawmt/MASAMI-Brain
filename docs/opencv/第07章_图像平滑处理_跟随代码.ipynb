## 均值滤波
`dst=cv2.blur(src,ksize,anchor,borederType)`
- `src`：输入图像
- `ksize`：卷积核大小，必须为奇数，如`ksize=(3,3)`  
- `anchor`：锚点，默认为`( -1, -1)`，表示锚点位于中心。
- `borderType`：边界填充方式，默认为`BORDER_DEFAULT`，表示根据边界情况决定填充方式。

```python
import cv2

img=cv2.imread("picture5_gaussian.jpg")

dst=cv2.blur(img,(3,3))
cv_show("original",img)
cv_show("blur",dst)
```

## 方框滤波
两种形式：
- 邻域像素值加权平均：用窗口内的像素值加权平均作为输出值
- 邻域像素值之和：用窗口内的像素值之和作为输出值
`dst=cv2.boxFilter(src,ddepth,ksizem,anchor,normalize,borderType)`
- `src`：输入图像
- `ddepth`：输出图像的深度，可以是`CV_8U`、`CV_16U`、`CV_16S`、`CV_32F`、`CV_64F`中的一种
- `ksizem`：滤波器的大小，是一个奇数
- `anchor`：锚点，默认为`( -1, -1 )`，表示锚点位于中心
- `normalize`：是否归一化，默认为`true`，表示归一化
- `borderType`：边界填充方式，可以是`BORDER_CONSTANT`、`BORDER_REPLICATE`、`BORDER_REFLECT`、`BORDER_WRAP`、`BORDER_REFLECT_101`中的一种

```python
import cv2

img=cv2.imread("picture3.jpg")

dst=cv2.boxFilter(img,-1,(3,3),normalize=1)

cv_show("original",img)
cv_show("boxFilter",dst)

```

## 高斯滤波
`dst=cv2.GaussianBlur(src,ksize,sigmaX,sigmaY,borderType)`
- `src`：输入图像   
- `ksize`：高斯核的大小，必须为奇数，如3,5,7等   
- `sigmaX`：X方向的标准差，如果为0，则根据ksize自动计算   
- `sigmaY`：Y方向的标准差，如果为0，则根据ksize自动计算   
- `borderType`：边界类型，默认值为cv2.BORDER_DEFAULT


```python
import cv2

img=cv2.imread("picture5_gaussian.jpg")

dst=cv2.GaussianBlur(img,(5,5),sigmaX=0,sigmaY=0)
cv_show("dst",dst)
```

## 中值滤波
`dst=cv2.medianBlur(src,ksize)`
- `src`：输入图像
- `ksize`：滤波器的大小，必须为奇数,1,3,5,7,9等


```python
import cv2

img=cv2.imread("picture1.jpg")

dst=cv2.medianBlur(img,9)
cv_show("dst",dst)
```

## 双边滤波
`dst = cv2.bilateralFilter(src, d, sigmaColor, sigmaSpace)`
- `src`：输入图像
- `d`：滤波器的领域半径
- `sigmaColor`：颜色空间的标准差
- `sigmaSpace`：坐标空间的标准差

```python
import cv2

img=cv2.imread("picture3.jpg")

dst=cv2.bilateralFilter(img,10,100,100)

cv_show("original",img)
cv_show("bilateralFilter",dst)
```

## 2D卷积

`dst=cv2.filter2D(src,ddepth,kernel,anchor=(-1,-1),delta=0,borderType=cv2.BORDER_DEFAULT)`
- `src`：输入图像，即待卷积图像。
- `ddepth`：输出图像的深度，可以为-1、CV_8U、CV_16U、CV_16S、CV_32F、CV_64F。
- `kernel`：卷积核，即卷积模板。
- `anchor`：锚点，即卷积核的中心位置，默认值为(-1,-1)，表示锚点位于模板中心。
- `delta`：位移量，即卷积结果的偏移量。
- `borderType`：边界填充方式，默认值为cv2.BORDER_DEFAULT。
